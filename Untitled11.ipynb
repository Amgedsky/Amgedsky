```python
import matplotlib.pyplot as plt

def run_paper_simulation():
    time_steps = 6
    levels = [1, 3, 5]  # المستوى 1 (النخبة)، 3 (الوسطى)، 5 (الدنيا)

    data = {}

    for lvl in levels:
        # تحديد الموارد والقيم الابتدائية بناءً على الموقع الهرمي
        if lvl == 1:
            res, val = 90.0, 85.0
        elif lvl == 3:
            res, val = 50.0, 40.0
        else:
            res, val = 20.0, 15.0

        history_val = [val]

        for t in range(1, time_steps):
            # قرار الفرد المرتبط بالموارد المتاحة D(i,t)
            decision = min(1.0, max(0.1, res / 100.0 * 0.9 + (6 - lvl) * 0.05))

            # المقاومة البيئية تزداد كلما اتجهنا نحو المستويات الأدنى
            opposition = 1.0 + (lvl * 0.4)

            # معادلة تطور الفرد وقيود الإنتروبيا g(...)
            potential_gain = (res * decision) / opposition
            entropy_pull = lvl * 0.35

            val = max(0.0, val + potential_gain - entropy_pull)

            # معادلة استنزاف الموارد h(...)
            consumption = decision * 12
            replenishment = 16 if lvl == 1 else (6 if lvl == 3 else 2)
            res = max(0.0, min(100.0, res - consumption + replenishment))

            history_val.append(val)

        data[lvl] = history_val

    # رسم النتائج باستخدام Matplotlib
    plt.figure(figsize=(9, 5), dpi=100)
    for lvl in levels:
        label_name = f"Level {lvl} (Elite)" if lvl == 1 else (f"Level {lvl} (Middle)" if lvl == 3 else f"Level {lvl} (Lower)")
        plt.plot(range(time_steps), data[lvl], marker='o', linewidth=2.5, label=label_name)

    plt.title("Simulation of Individual State Evolution Across Hierarchy Levels", fontsize=12, fontweight='bold', pad=12)
    plt.xlabel("Time Steps (t)", fontsize=10)
    plt.ylabel("State Value V(s)", fontsize=10)
    plt.xticks(range(time_steps))
    plt.legend()
    plt.tight_layout()
    plt.savefig("simulation_plot.png")
    plt.show()

if __name__ == "__main__":
    run_paper_simulation()
```
